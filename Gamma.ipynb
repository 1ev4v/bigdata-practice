```python
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/7_1_1.csv')
# print(df.head())
# print(df.info())

df.loc[:, ['신고일자','신고시각','처리일자','처리시각']] = df.loc[:, ['신고일자','신고시각','처리일자','처리시각']].astype(str)

df['신고시각'] = df['신고시각'].str.zfill(6)
df['처리시각'] = df['처리시각'].str.zfill(6)

df['신고일시'] = pd.to_datetime(df['신고일자'] + df['신고시각'], format = '%Y%m%d%H%M%S')
df['처리일시'] = pd.to_datetime(df['처리일자'] + df['처리시각'], format = '%Y%m%d%H%M%S')

df['처리시간'] = (df['처리일시'] - df['신고일시']).dt.total_seconds()
df.groupby('공장명')['처리시간'].mean().sort_values(ascending=False).index[2]
df
```

```python
import numpy as np
import pandas as pd

train_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/7_2_train.csv')
test_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/7_2_test.csv')

train_X = train_df.drop('Target', axis=1)
train_y = train_df['Target']
test_X = test_df.drop('Target', axis=1)
test_y = test_df.pop('Target')

# train_X.info()

from sklearn.model_selection import train_test_split

train_X, valid_X, train_y, valid_y = train_test_split(train_X, train_y, test_size=0.3)
# print(train_X.shape, valid_X.shape, train_y.shape, valid_y.shape)
# print(dir(train_X.select_dtypes('object').columns))
cat_train_X = train_X.select_dtypes('object').columns.to_list()
num_train_X = train_X.select_dtypes('number').columns.to_list()
cat_valid_X = valid_X.select_dtypes('object').columns.to_list()
num_valid_X = valid_X.select_dtypes('number').columns.to_list()
cat_test_X = test_X.select_dtypes('object').columns.to_list()
num_test_X = test_X.select_dtypes('number').columns.to_list()

from sklearn.preprocessing import OneHotEncoder
# print(dir(sklearn.preprocessing))
# print(help(OneHotEncoder))
onehotencoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
# print(dir(onehotencoder))
onehot_cat_train_X = onehotencoder.fit_transform(train_X[cat_train_X])
onehot_cat_valid_X = onehotencoder.transform(valid_X[cat_valid_X])
onehot_cat_test_X = onehotencoder.transform(test_X[cat_test_X])

train_X1 = np.concatenate([onehot_cat_train_X, train_X[num_train_X]], axis=1)
valid_X1 = np.concatenate([onehot_cat_valid_X, valid_X[num_valid_X]], axis=1)
test_X1 = np.concatenate([onehot_cat_test_X, test_X[num_test_X]], axis=1)

from sklearn.ensemble import GradientBoostingClassifier
rf = GradientBoostingClassifier()
rf.fit(train_X1, train_y)

# pred1 = rf.predict(valid_X1)
pred2 = rf.predict(test_X1)

from sklearn.metrics import f1_score
# print(help(f1_score))
# rs1 = f1_score(valid_y, pred1, average='macro')
rs2 = f1_score(test_y, pred2, average='macro')
print(rs2)

pred2 = pd.DataFrame(pred2, columns=['pred'])
pred2.to_csv('result.csv', index=False)
```

```python
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/7_3_1.csv')

# print(df.head())
# print(df.info())
print(np.round(df['before'].mean(),2))
print(np.round(df['after'].mean(),2))
print(np.round(df['before'].std(),2))
print(np.round(df['after'].std(),2))

from scipy.stats import ttest_rel
# print(help(ttest_rel))
statistic, pvalue = ttest_rel(df['before'], df['after'])
print(np.round(statistic,2))
print(np.round(pvalue,3) < 0.05)
```

```python
import numpy as np
import pandas as pd

train_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/7_3_2_train.csv')
test_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/7_3_2_test.csv')

# print(train_df.info())
# print(test_df.info())

# train_df.head()

import statsmodels.formula.api as smf
import statsmodels.api as sm
model = smf.glm("purchase ~ age + income + family_members", data=train_df, family=sm.families.Binomial()).fit()
# print(model.summary())

# model.predict()

# print(dir(model))
print(np.exp(model.params['income']))
print(model.deviance)

# print(dir(sklearn.metrics))
test_X = test_df.drop('purchase', axis=1)
test_y = test_df['purchase']

# print(test_X.shape, test_y.shape)

pred = model.predict(test_X)
pred = (pred >= 0.5).astype(int)
rs = np.mean(pred != test_y)
print(rs)
```
