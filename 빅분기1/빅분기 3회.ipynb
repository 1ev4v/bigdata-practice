```python
# 1-3
import numpy as np
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/3_1.csv')
# print(df.info())
# print(df.head())

up = df['MedianStudentsPerTeacher'].mean() + (1.5 * df['MedianStudentsPerTeacher'].std())
down = df['MedianStudentsPerTeacher'].mean() - (1.5 * df['MedianStudentsPerTeacher'].std())

# print(up, down)

print(df[(df['MedianStudentsPerTeacher'] > up) | (df['MedianStudentsPerTeacher'] < down)]['MedianStudentsPerTeacher'].count())
```

```python
# 2
import numpy as np
import pandas as pd

train_X = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/3_2_trainX.csv')
train_y = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/3_2_trainy.csv')
test_X = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/3_2_testX.csv')

print(train_X.info())
print(train_X.head())
# print(train_y.shape, test_X.shape)

from sklearn.model_selection import train_test_split
train_X, valid_X, train_y, valid_y = train_test_split(train_X, train_y, test_size=0.3)

# print(train_X.shape, train_y.shape, valid_X.shape, valid_y.shape)

from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier()
rf.fit(train_X, train_y)
pred1 = rf.predict(valid_X)

from sklearn.metrics import f1_score
print(f1_score(pred1, valid_y))

pred2 = rf.predict(test_X)
pd.DataFrame(pred2, columns=['pred']).to_csv('result.csv', index=False)

from sklearn.model_selection import GridSearchCV
params = {
    'max_depth': [10, 20, 30],
    'min_samples_split': [2, 5, 10]
}

from sklearn.metrics import get_scorer_names
# print(get_scorer_names())

gridcv = GridSearchCV(estimator=rf, param_grid=params, cv=3, scoring='f1')
gridcv.fit(train_X, train_y)
pred3 = gridcv.predict(valid_X)
print(f1_score(pred3, valid_y))

```

```python
# 3-1
import numpy as np
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/3_3_1.csv')
# print(df.info())
# print(df.head())

print(df['score'].mean().round(2))
print(np.round(df['score'].std(), 2))

from scipy.stats import ttest_1samp
statistic, pvalues = ttest_1samp(df['score'], popmean=75)
print(statistic.round(2))
print(pvalues < 0.05)
```

```python
# 3-2
import numpy as np
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/3_3_2.csv')
# print(df.info())
# print(df.head())

df1 = pd.crosstab(df['gender'], df['club_membership'])

from scipy.stats import chi2_contingency
res = chi2_contingency(df1, correction=False)
print(res.statistic.round(2))
print(res.pvalue < 0.05)
```
