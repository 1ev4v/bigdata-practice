```python
# 2
import numpy as np
import pandas as pd
train_df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/5_2_train.csv')
test_df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/5_2_test.csv')
# print(train_df.info())
# print(train_df.head())
# print(test_df.info())

train_y = train_df['target']
train_X = train_df.drop(columns=['level_0', 'ID', 'target'], axis=1)
test_y = test_df['target']
test_X = test_df.drop(columns=['level_0', 'ID', 'target'], axis=1)

from sklearn.metrics import roc_auc_score, get_scorer_names
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
# roc_auc

train_X, valid_X, train_y, valid_y = train_test_split(train_X, train_y, test_size=0.3)
# print(train_X.shape, train_y.shape, valid_X.shape, valid_y.shape)
# print(np.isnan(train_X).sum())

imputer = SimpleImputer(strategy='mean')
train_X = imputer.fit_transform(train_X)
valid_X = imputer.transform(valid_X)
test_X = imputer.transform(test_X)
# print(np.isnan(train_X).sum())

rf=RandomForestClassifier()
rf.fit(train_X, train_y)
pred1 = rf.predict_proba(valid_X)[:,1]

print(roc_auc_score(valid_y, pred1))

# train_y.value_counts()
# valid_y.value_counts()
# rf.feature_importances_

params={
    'max_depth': [10,20,30],
    'min_samples_split': [2,5,10]
}
cv=GridSearchCV(estimator=rf, param_grid=params, scoring='roc_auc', cv=3)
cv.fit(train_X, train_y)
pred2 = cv.predict_proba(valid_X)[:,1]
print(roc_auc_score(valid_y, pred2))

pred3 = cv.predict_proba(test_X)[:,1]

pd.DataFrame(pred3, columns=['pred']).to_csv('result.csv', index=False)
```

```python
# 3-1
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/5_3_1.csv')
print(df.info())
print(df.head())

print(np.round(df['study_hours'].mean(), 2))
print(np.round(df['study_hours'].std(), 2))
print(np.round(df['exam_scores'].mean(), 2))
print(np.round(df['exam_scores'].std(), 2))

from scipy.stats import pearsonr
print(pearsonr(df['study_hours'], df['exam_scores']).statistic.round(3))
print(pearsonr(df['study_hours'], df['exam_scores']).pvalue.round(4) > 0.05)
```

```python
# 3-2
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/5_3_2.csv')
# print(df.info())
# print(df.head())

a = df.groupby(['campaign']).mean()
print((a.iloc[0] + a.iloc[1] + a.iloc[2]).round(2))

b = df.groupby(['campaign']).std()
print((b.iloc[0] + b.iloc[1] + b.iloc[2]).round(2))

from scipy.stats import f_oneway

A = df[df['campaign'] == 'A']['satisfaction_score']
B = df[df['campaign'] == 'B']['satisfaction_score']
C = df[df['campaign'] == 'C']['satisfaction_score']

print(f_oneway(A,B,C).statistic.round(3))
print(f_oneway(A,B,C).pvalue.round(4) > 0.05)
```
