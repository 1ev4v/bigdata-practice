```python
# 1-1
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/5_1_1.csv')
# print(df.info())
# print(df.head())
df['conventional']=df['conventional'].replace(r'[^0-9]', None, regex=True)
df=df.dropna()
df['conventional']=df['conventional'].astype(int)
# print(df.head())
# print(df.info())
print(np.round(df['conventional'].quantile(0.75) - df['conventional'].quantile(0.25),0))
```
