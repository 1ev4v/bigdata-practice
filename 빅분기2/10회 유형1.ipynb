```python
# 1-2
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/10_1_2.csv')
# print(df.info())
# print(df.head())

df['year']=df['date'].str.split('-').str[0]
df['month']=df['date'].str.split('-').str[1]
print(df.groupby(['year', 'month'])['price'].sum().sort_values(ascending=False)[1])
df.groupby(['year', 'month'])['price'].sum().sort_values(ascending=False).index[3]
print(df[(df['year']=='2024') & (df['month']=='12')].groupby('category')['price'].sum().sort_values(ascending=False)[0])
```

```python
# 1-3
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/10_1_3.csv')
# print(df.info())
# print(df.head())

df['split']=df['message'].str.split(' ')
def check(x):
    return len(x)
df['len']=df['split'].apply(check)
a=df.groupby('label')['len'].mean()
print(np.round(np.abs(a[0] - a[1]) ,3))
```
