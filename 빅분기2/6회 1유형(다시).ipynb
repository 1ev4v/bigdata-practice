```python
# 1-1
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/6_1_1.csv')
# print(df.info())
# print(df.head())

df=df[(df['ProductA가격'] != 0) & (df['ProductB가격'] != 0)]
# print(df.info())
# print(df.head())

df['차이']=np.abs(df['ProductA가격'] - df['ProductB가격'])
print(np.round(df.groupby('도시명')['차이'].mean().sort_values(ascending=False).values[0], 1))
```

```python
# 1-2
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/6_1_2.csv')
print(df.info())
print(df.head())

df['BMI']=df['Weight_kg']/((df['Height_cm']/100)**2)
df
def check(x):
    if x < 18.5:
        return '저체중'
    elif (x >= 18.5) & (x < 23):
        return '정상'
    elif (x >= 23) & (x < 25):
        return '과체중'
    else:
        return '비만'
df['rs']=df['BMI'].apply(check)
df
len(df[(df['rs']=='저체중') | (df['rs']=='비만')])
```
