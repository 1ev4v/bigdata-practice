```python
# 기출문제3회 1-2
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/3_1.csv')
# print(df.info())
# print(df.head())
a=df['numOfBathrooms'].std()
df['numOfBathrooms'].fillna(df['numOfBathrooms'].mean(), inplace=True)
b=df['numOfBathrooms'].std()
# print(df.info())

print(np.round(np.abs(a-b),2))
```
