```python
# 1-1
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_1_1.csv')
# print(df.info())
# print(df.head())

print(df.groupby('대륙')['맥주소비량'].mean().sort_values(ascending=False))
print(df[df['대륙']=='SA'].groupby('국가')['맥주소비량'].sum().sort_values(ascending=False).index[4])
print(np.round(df[df['국가']=='Venezuela']['맥주소비량'].mean(), 0))
```

```python
# 1-2
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_1_2.csv')
# print(df.info())
# print(df.head())

df1=df.groupby('국가')[['관광', '사무', '공무', '유학', '기타']].sum().reset_index()
df1['합계']=df1['관광'] + df1['사무'] + df1['공무'] + df1['유학'] + df1['기타']
df1['관광객비율']=df1['관광']/df1['합계']
print(df1.loc[df1['관광객비율'].sort_values(ascending=False).index[1]]['관광'])
# df.loc[df['관광'].sort_values(ascending=False).index[1]]['국가']
print(np.round(df[df['국가']=='이스라엘']['공무'].mean(),0))
print(9039+494)
```

```python
# 1-3
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_1_3.csv')
# print(df.info())
# print(df.head())

from sklearn.preprocessing import minmax_scale
df['CO(GT)']=minmax_scale(df['CO(GT)'])
df['NMHC(GT)']=minmax_scale(df['NMHC(GT)'])
# df.head()
print(np.round(df['CO(GT)'].std(),2))
print(np.round(df['NMHC(GT)'].std(),2))
```
