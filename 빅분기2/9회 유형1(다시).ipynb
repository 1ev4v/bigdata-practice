```python
# 1-1
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/9_1_1.csv')
# print(df.info())
# print(df.head())

df['합계']=df['금액1']+df['금액2']
df1=df.groupby(['year', '지역코드', 'gender'])['합계'].sum().reset_index()
df2=df1.pivot(index=['year', '지역코드'], columns='gender', values='합계')
df2=df2.fillna(0)
df2['차이']=np.abs(df2[0] - df2[1])
df2['차이'].sort_values(ascending=False).index[0][1]
```

```python
# 1-2
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/9_1_2.csv')
# print(df.info())
# print(df.head())

df1=df[df['구분']=='발생건수'].drop(columns=['구분']).set_index('연도')
df2=df[df['구분']=='검거건수'].drop(columns=['구분']).set_index('연도')
df3=df2/df1
df3
def check(x):
    return x == x.max()
df4=df3.apply(check)
df4
print(df2[df4].fillna(0).sum().sum())
```

```python
# 1-3
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/9_1_3.csv')
# print(df.info())
# print(df.head())

df['평균만족도']=df['평균만족도'].fillna(df['평균만족도'].mean())
df['근속연수']=df['근속연수'].fillna(np.floor(df.groupby(['부서', '등급'])['근속연수'].transform('mean')))
# print(df.info())
# print(df.head())

a=df[(df['부서']=='HR') & (df['등급']=='A')]['근속연수'].mean()
b=df[(df['부서']=='Sales') & (df['등급']=='B')]['교육참가횟수'].mean()
print(a+b)
```
