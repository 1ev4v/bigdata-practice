```python
# 1-1
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/7_1_1.csv')
print(df.info())
print(df.head())

df['처리시각']=df['처리시각'].astype(str)
df['신고시각']=df['신고시각'].astype(str)
df['처리시각']=df['처리시각'].str.zfill(6)
df['신고시각']=df['신고시각'].str.zfill(6)
df['신고일자']=df['신고일자'].astype(str)
df['처리일자']=df['처리일자'].astype(str)
df['신고일시']=df['신고일자'] + df['신고시각']
df['처리일시']=df['처리일자'] + df['처리시각']
df['신고일시']=pd.to_datetime(df['신고일시'],format='%Y%m%d%H%M%S')
df['처리일시']=pd.to_datetime(df['처리일시'],format='%Y%m%d%H%M%S')
# df['처리시각']=df['처리시각'].astype(int)
# df['신고시각']=df['신고시각'].astype(int)
df['처리일시']=(df['처리일시'] - df['신고일시'])
df.groupby('공장명')['처리일시'].mean().sort_values()
```

```python
# 1-3
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/7_1_3.csv')
# print(df.info())
# print(df.head())

df['year']=df['기간'].str.split('_').str[0]
df['month']=df['기간'].str.split('_').str[1].str.replace('월','')
df['총판매량']=df['제품A'] + df['제품B'] + df['제품C'] + df['제품D'] + df['제품E']
# df.info()
def check(x):
    if x in ['1','2','3']:
        return 1
    elif x in ['4','5','6']:
        return 2
    elif x in ['7','8','9']:
        return 3
    else:
        return 4
df['분기']=df['month'].apply(check)
(df.groupby(['year', '분기'])['총판매량'].sum() / 3).sort_values(ascending=False)
```
