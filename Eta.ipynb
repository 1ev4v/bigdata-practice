```python
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/10_1_1.csv')

print(df.info())
print(df.head())

df1 = (df.groupby('소주제')['정답여부'].sum() / df.groupby('소주제')['정답여부'].count())
# print(help(df1.sort_values(ascending=False).rank(ascending=False, method='dense')))
df1[df1.index==df1.sort_values(ascending=False).rank(ascending=False, method='dense').index[3]]
```

```python
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/10_1_2.csv')

print(df.info())
print(df.head())

df['year'] = df['date'].str.split('-').str[0]
df['month'] = df['date'].str.split('-').str[1]
rs = df.groupby(['year', 'month'])['price'].sum().sort_values(ascending=False).index[3]
df[(df['year']==rs[0]) & (df['month']==rs[1])].groupby('category')['price'].sum().sort_values(ascending=False)
```
