```python
import numpy as np
import pandas as pd

# dir(pd)

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/3_1.csv')

# print(df.head())
# print(df.info())

df1 = df.sort_values(by='lotSizeSqFt', ascending=False).reset_index()
df1.loc[:9,'lotSizeSqFt'] = df1.loc[9, 'lotSizeSqFt']
print(f'### 1.1번: {np.round(df1[df1['yearBuilt'] >= 2000]['lotSizeSqFt'].mean(),0)}')

alpha = df1['numOfBathrooms'].std()
beta = df1['numOfBathrooms'].fillna(df1['numOfBathrooms'].median()).std()
print(f'### 1.2번: {np.round(alpha - beta, 2)}')

print(f'### 1.3번: {df1[df1['MedianStudentsPerTeacher'] >= df1['MedianStudentsPerTeacher'].mean() + (1.5 * df1['MedianStudentsPerTeacher'].std())]['MedianStudentsPerTeacher'].count() + df1[df1['MedianStudentsPerTeacher'] <= df1['MedianStudentsPerTeacher'].mean() - (1.5 * df1['MedianStudentsPerTeacher'].std())]['MedianStudentsPerTeacher'].count()}')
```

```python
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/3_3_1.csv')
df.head()
print(df.info())

print(f'### 3-1-1번: {np.round(df['score'].mean(),2)}')
print(f'### 3-1-2번: {np.round(df['score'].std(),2)}')

from scipy.stats import ttest_1samp
# help(ttest_1samp)
rs = ttest_1samp(df['score'], popmean=75)
print(f'### 3-1-3번: {rs.statistic.round(2)}')
# print(rs.pvalue.round(2))

if rs.pvalue.round(2) < 0.05:
    alpha = '기각'
else:
    alpha = '채택'

print(f'### 3-1-4번: {alpha}')
```
