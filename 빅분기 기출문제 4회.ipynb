```python
# 3-1-3
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/4_3_1.csv')
# df.head()
# df.info()
A = df[df['department'] == 'A']['hours_worked']
B = df[df['department'] == 'B']['hours_worked']

# from scipy import stats
# print(dir(stats))

from scipy.stats import ttest_ind
# print(help(ttest_ind))
statistic, p_value = ttest_ind(A, B, equal_var=True)
print(round(statistic, 2))
```

```python
# 3-2-1
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/4_3_2.csv')
# df.head()
# df.info()
df['factory'].value_counts()
A = df[df['factory'] == 'A']['quality_score'].mean()
B = df[df['factory'] == 'B']['quality_score'].mean()
C = df[df['factory'] == 'C']['quality_score'].mean()
rs = round(A + B + C, 2)
rs
```

```python
# 3-2-3
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/4_3_2.csv')
A = df[df['factory'] == 'A']['quality_score']
B = df[df['factory'] == 'B']['quality_score']
C = df[df['factory'] == 'C']['quality_score']

from scipy.stats import kruskal
statistic, p_value = kruskal(A, B, C)
print(round(statistic, 2))
```
