```python
# 1
import pandas as pd
import numpy as np

np.random.seed(42)
n_samples = 210
X = np.random.randn(n_samples, 4)
y = (X[:, 0] + X[:, 1] * 0.5 + np.random.randn(n_samples) * 0.5 > 0).astype(int)
df = pd.DataFrame(X, columns=['weight', 'height', 'age', 'income'])
df['gender'] = y

# print(df.info())
# print(df.head())

import statsmodels.api as sm
import statsmodels.formula.api as smf

model = smf.glm("gender ~ weight", data=df, family=sm.families.Binomial()).fit()
# print(model.summary())
print(f'오즈비: {np.exp(model.params["weight"])}')
```

```python
# 3
import pandas as pd
import numpy as np

np.random.seed(42)
n_samples = 210
X = np.random.randn(n_samples, 4)
y = (X[:, 0] + X[:, 1] * 0.5 + np.random.randn(n_samples) * 0.5 > 0).astype(int)
df = pd.DataFrame(X, columns=['weight', 'height', 'age', 'income'])
df['gender'] = y

# print(df.info())
# print(df.head())

df = df.drop(['height', 'age', 'income'], axis=1)

from sklearn.model_selection import train_test_split
df_train, df_test = train_test_split(df, test_size=90, random_state=42)
# print(df_train.shape, df_test.shape)
print(df_train.info())
print(df_test.info())

import statsmodels.api as sm 
import statsmodels.formula.api as smf

model = smf.glm("gender ~ weight", data=df_train, family=sm.families.Binomial()).fit()
test_X = df_test.drop('gender', axis=1)
test_y = df_test['gender']
rs = model.predict(test_X)
rs = (rs > 0.5).astype(int)
from sklearn.metrics import accuracy_score
final_rs = np.round(1 - accuracy_score(test_y, rs), 3)
print(f'오분류율: {final_rs}')
```
