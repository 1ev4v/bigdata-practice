```python
# 1-2 
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_1_2.csv')
df1 = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_1_2.csv')
# print(df.info())
# print(df.head())

df = df.groupby('국가').sum().reset_index()

df['합계'] = df['관광'] + df['사무'] + df['공무'] + df['유학'] + df['기타']
df['관광객비율'] = np.round(df['관광'] / df['합계'], 3)
df1['합계'] = df1['관광'] + df1['사무'] + df1['공무'] + df1['유학'] + df1['기타']
df1['관광객비율'] = np.round(df1['관광'] / df1['합계'], 3)

rs1 = df[df['국가'] == df[df['관광객비율'] == df['관광객비율'].sort_values(ascending=False).values[1]]['국가'].values[0]]['관광'].sum()
print(rs1)

rs2 = np.round(df1[df1['국가'] == df1[df1['관광'] == df1['관광'].sort_values(ascending=False).values[1]]['국가'].values[0]]['공무'].mean(), 0)
print(rs2)

print(np.round(rs1 + rs2))
```

```python
# 2
import numpy as np
import pandas as pd

train_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_2_train.csv')
test_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_2_test.csv')

print(train_df.info())
print(train_df.head())
# print(test_df.info())
# print(test_df.head())

train_y = train_df['count']
train_X = train_df.drop(columns=['count', 'ID'])
test_y = test_df['count']
test_X = test_df.drop(columns=['count', 'ID'])

from sklearn.model_selection import train_test_split
train_X, valid_X, train_y, valid_y = train_test_split(train_X, train_y, test_size=0.3)

train_X_cat = train_X.select_dtypes('object').columns.tolist()
valid_X_cat = valid_X.select_dtypes('object').columns.tolist()
test_X_cat = test_X.select_dtypes('object').columns.tolist()
train_X_num = train_X.select_dtypes('number').columns.tolist()
valid_X_num = valid_X.select_dtypes('number').columns.tolist()
test_X_num = test_X.select_dtypes('number').columns.tolist()

from sklearn.preprocessing import OneHotEncoder
onehotencoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
one_train_X = onehotencoder.fit_transform(train_X[train_X_cat])
one_valid_X = onehotencoder.transform(valid_X[valid_X_cat])
one_test_X = onehotencoder.transform(test_X[test_X_cat])

train_X = np.concatenate([train_X[train_X_num], one_train_X], axis=1)
valid_X = np.concatenate([valid_X[valid_X_num], one_valid_X], axis=1)
test_X = np.concatenate([test_X[test_X_num], one_test_X], axis=1)

from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor()

rf.fit(train_X, train_y)
pred1 = rf.predict(valid_X)

from sklearn.metrics import mean_absolute_error
print(mean_absolute_error(valid_y, pred1))

pred = rf.predict(test_X)
pd.DataFrame(pred, columns=['pred']).to_csv('result.csv', index=False)
```

```python
# 3-1
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_3_1.csv')
# print(df.info())
# print(df.head())

print(np.round(df['before'].mean(), 2))
print(np.round(df['before'].std(), 2))
print(np.round(df['after'].mean(), 2))
print(np.round(df['after'].std(), 2))

df['차이'] = df['after'] - df['before']

from scipy.stats import wilcoxon
statistic, p_value = wilcoxon(df['before'], df['after'])
print(np.round(statistic, 2))
print(np.round(p_value, 3) > 0.05)
```

```python
# 3-2
import numpy as np
import pandas as pd

train_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_3_2_train.csv')
test_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_3_2_test.csv')

# print(train_df.info())
# print(train_df.head())
# print(test_df.info())

import statsmodels.api as sm
import statsmodels.formula.api as smf

model = smf.ols("productivity ~ hours + age + experience", data=train_df).fit()
# print(model.summary())
print(model.params[1:].idxmax())
print((model.pvalues[1:] < 0.05).sum())

test_y = test_df['productivity']
test_X = test_df.drop(columns=['productivity'])
pred = model.predict(test_X)

from sklearn.metrics import r2_score
print(np.floor((r2_score(test_y, pred) * 100)))
```
