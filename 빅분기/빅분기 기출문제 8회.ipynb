```python
# 1-2
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_1_2.csv')
print(df.info())
print(df.head())

df['합계'] = df['관광'] + df['사무'] + df['공무'] + df['유학'] + df['기타']
df['관광객비율'] = np.round(df['관광'] / df['합계'], 3)
rs1 = df[df['관광객비율'] == df['관광객비율'].sort_values(ascending=False).values[1]]['국가'].values[0]
rs11 = df[df['국가'] == rs1]['관광'].sum()
rs2 = np.round(df[df['국가'] == rs1]['공무'].mean(), 0)
print(rs11)
print(rs2)
print((rs11 + rs2).astype(int))
```

```python
# 1-3
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_1_3.csv')
# print(df.info())
# print(df.head())

from sklearn.preprocessing import minmax_scale
df['CO(GT)'] = minmax_scale(df['CO(GT)'])
df['NMHC(GT)'] = minmax_scale(df['NMHC(GT)'])

print(np.round(df['CO(GT)'].std(), 2))
print(np.round(df['NMHC(GT)'].std(), 2))
```

```python
# 3-1
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_3_1.csv')
# print(df.info())
# print(df.head())

print(np.round(df['before'].mean(), 2))
print(np.round(df['before'].std(), 2))
print(np.round(df['after'].mean(), 2))
print(np.round(df['after'].std(), 2))

df['차이'] = df['after'] - df['before']

from scipy.stats import wilcoxon
statistic, p_value = wilcoxon(df['before'], df['after'])
print(np.round(statistic, 2))
print(np.round(p_value, 3) > 0.05)
```

```python
# 3-2
import numpy as np
import pandas as pd

train_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_3_2_train.csv')
test_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_3_2_test.csv')

# print(train_df.info())
# print(train_df.head())
# print(test_df.info())

import statsmodels.api as sm
import statsmodels.formula.api as smf

model = smf.ols("productivity ~ hours + age + experience", data=train_df).fit()
# print(model.summary())
print(model.params[1:].idxmax())
print((model.pvalues[1:] < 0.05).sum())

test_y = test_df['productivity']
test_X = test_df.drop(columns=['productivity'])
pred = model.predict(test_X)

from sklearn.metrics import r2_score
print(np.floor((r2_score(test_y, pred) * 100)))
```
