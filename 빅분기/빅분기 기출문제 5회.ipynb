```python
# 1-1   x
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/5_1_1.csv')
# print(df.head())
# print(df.info())

df['conventional'] = df['conventional'].str.replace(r'[^0-9]', '', regex=True)
df['conventional'] = pd.to_numeric(df['conventional'], errors='coerce')
# df['conventional']
df['conventional'] = df['conventional'].fillna(0).astype(int)
df['conventional'] = df['conventional'].astype(int)

rs = round(df['conventional'].quantile(0.75) - df['conventional'].quantile(0.25), 0)
print(rs)
```

```python
# 3-2-3
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/5_3_2.csv')
# print(df.head())
# print(df.info())

alpha = {}
for i in df['campaign'].unique():
    alpha[i] = df[df['campaign'] == i]['satisfaction_score']

from scipy.stats import f_oneway
statistic, p_value = f_oneway(alpha['A'], alpha['B'], alpha['C'])
print(round(statistic, 3))
```
