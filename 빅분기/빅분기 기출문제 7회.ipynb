```python
# 1-3
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/7_1_3.csv')
# print(df.info())
# print(df.head())

df['년도'] = df['기간'].str.split('_').str[0]
df['월'] = df['기간'].str.split('_').str[1].str.replace('월', '').astype(int)
df['분기'] = df['월'].apply(lambda x: 1 if x in [1, 2, 3] else (2 if x in [4, 5, 6] else (3 if x in [7, 8, 9] else 4)))
df['총 제품'] = df[['제품A', '제품B', '제품C', '제품D', '제품E']].sum(axis=1)
np.round(df.groupby(['년도', '분기'])['총 제품'].mean().max(), 6)
# df.groupby(['년도', '분기'])['총 제품'].mean().idxmax()
```

```python
# 2
import numpy as np
import pandas as pd

train_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/7_2_train.csv')
test_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/7_2_test.csv')

test_id = test_df['ID'] 

# print(train_df.info())
# print(train_df.head())
# print(test_df.info())

train_df = train_df.drop(columns=['level_0', 'ID'])
test_df = test_df.drop(columns=['level_0', 'ID'])

train_y = train_df['Target']
train_X = train_df.drop(columns=['Target'])
test_y = test_df['Target']
test_X = test_df.drop(columns=['Target'])

# print(train_y.head())
# print(train_X.head())
# print(train_X.info())

# print(train_X.shape, train_y.shape, test_X.shape, test_y.shape)

from sklearn.model_selection import train_test_split
train_X, valid_X, train_y, valid_y = train_test_split(train_X, train_y, test_size=0.3)
# print(train_X.shape, valid_X.shape, train_y.shape, valid_y.shape)

train_X_cat = train_X.select_dtypes('object').columns.tolist()
train_X_num = train_X.select_dtypes('number').columns.tolist()
valid_X_cat = valid_X.select_dtypes('object').columns.tolist()
valid_X_num = valid_X.select_dtypes('number').columns.tolist()
test_X_cat = test_X.select_dtypes('object').columns.tolist()
test_X_num = test_X.select_dtypes('number').columns.tolist()

train_X[train_X_cat] = train_X[train_X_cat].replace('_', '', regex=True)
valid_X[valid_X_cat] = valid_X[valid_X_cat].replace('_', '', regex=True)
test_X[test_X_cat] = test_X[test_X_cat].replace('_', '', regex=True)

# print(train_X.info())
# print(train_X.head())

train_X_cat = train_X.select_dtypes('object').columns.tolist()
train_X_num = train_X.select_dtypes('number').columns.tolist()
valid_X_cat = valid_X.select_dtypes('object').columns.tolist()
valid_X_num = valid_X.select_dtypes('number').columns.tolist()
test_X_cat = test_X.select_dtypes('object').columns.tolist()
test_X_num = test_X.select_dtypes('number').columns.tolist()

# print(train_X_cat)
# print(train_X_num)
# print(test_X_cat)
# print(train_X.info())
# print(test_X.info())


from sklearn.preprocessing import OneHotEncoder, LabelEncoder, OrdinalEncoder, StandardScaler
# onehotencoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
ordinalencoder = OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1)
stdscaler = StandardScaler()

# one_train_X_cat = onehotencoder.fit_transform(train_X[train_X_cat])
# one_valid_X_cat = onehotencoder.transform(valid_X[valid_X_cat])
# one_test_X_cat = onehotencoder.transform(test_X[test_X_cat])

one_train_X_cat = ordinalencoder.fit_transform(train_X[train_X_cat])
one_valid_X_cat = ordinalencoder.transform(valid_X[valid_X_cat])
one_test_X_cat = ordinalencoder.transform(test_X[test_X_cat])

scaler_train_X_num = stdscaler.fit_transform(train_X[train_X_num])
scaler_valid_X_num = stdscaler.transform(valid_X[valid_X_num])
scaler_test_X_num = stdscaler.transform(test_X[test_X_num])

train_X = np.concatenate([one_train_X_cat, scaler_train_X_num], axis=1)
valid_X = np.concatenate([one_valid_X_cat, scaler_valid_X_num], axis=1)
test_X = np.concatenate([one_test_X_cat, scaler_test_X_num], axis=1)

target_mapping = {'Dropout': 0, 'Enrolled': 1, 'Graduate': 2}  

train_y = train_y.map(target_mapping)
valid_y = valid_y.map(target_mapping)
test_y = test_y.map(target_mapping)

train_X_full = np.concatenate([train_X, valid_X], axis=0)
train_y_full = np.concatenate([train_y, valid_y], axis=0)

# le = LabelEncoder()

# train_y = le.fit_transform(train_y)
# valid_y = le.transform(valid_y)
# test_y = le.transform(test_y)

# print(train_X.shape, valid_X.shape, test_X.shape)

from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
rf = RandomForestClassifier()
# rf = GradientBoostingClassifier()

rf.fit(train_X, train_y)
# rf.fit(train_X_full, train_y_full)
pred1 = rf.predict(valid_X)

from sklearn.metrics import f1_score
print(f1_score(valid_y, pred1, average="macro"))

pred = rf.predict(test_X)
print(f1_score(test_y, pred, average="macro"))
pd.DataFrame(pred, columns=['pred']).to_csv('result.csv', index=False)

# result = pd.DataFrame({
#     'ID': test_id, 
#     'pred': pred     
# })
# result.to_csv('result.csv', index=False)
```

```python
# 3-2
import numpy as np
import pandas as pd

train_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/7_3_2_train.csv')
test_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/7_3_2_test.csv')

# print(train_df.head())
# print(train_df.info())

import statsmodels.api as sm
import statsmodels.formula.api as smf

model = smf.glm("purchase ~ age + income + family_members", data=train_df, family=sm.families.Binomial()).fit()
# print(model.summary())
print(np.round(np.exp(model.params['income']),3))
print(np.round(model.deviance, 2))

# print(test_df.info())
test_y = test_df['purchase']
test_X = test_df.drop(columns=['purchase'])
pred = model.predict(test_X)

pred = (pred >= 0.5).astype(int)

from sklearn.metrics import accuracy_score
print(1 - accuracy_score(test_y, pred))
```
