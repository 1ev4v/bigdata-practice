```python
# 1
import pandas as pd
import numpy as np

np.random.seed(42)
n_samples = 100
X = np.random.randn(n_samples, 5)
y = 3 * X[:, 0] + 2 * X[:, 1] + X[:, 2] + np.random.randn(n_samples)
df = pd.DataFrame(X, columns=['var1', 'var2', 'var3', 'var4', 'var5'])
df['target'] = y

# print(df.head())

rs1 = df.corr()
rs2 = rs1['target'].drop('target')
print(f'컬럼이름: {rs2.idxmax()}, 수치: {rs2.max()} ')
```

```python
# 2
import pandas as pd
import numpy as np

np.random.seed(42)
n_samples = 100
X = np.random.randn(n_samples, 5)
y = 3 * X[:, 0] + 2 * X[:, 1] + X[:, 2] + np.random.randn(n_samples)
df = pd.DataFrame(X, columns=['var1', 'var2', 'var3', 'var4', 'var5'])
df['target'] = y

# print(df.info())
# print(df.head())

import statsmodels.api as sm
import statsmodels.formula.api as smf
model = smf.ols("target ~ var1 + var2 + var3 + var4 + var5", data=df).fit()
# print(model.summary())
print(f'결정계수: {model.rsquared}')
```

```python
# 4
import pandas as pd
import numpy as np
from sklearn.datasets import make_regression
import statsmodels.api as sm

X, y = make_regression(n_samples=100, n_features=3, noise=0.1, random_state=42)
df = pd.DataFrame(X, columns=[f'var{i}' for i in range(3)])
df['target'] = y

# print(df.info())
# print(df.head())

import statsmodels.formula.api as smf
model = smf.ols("target ~ var0 + var1 + var2", data=df).fit()
# print(model.summary())
print(f'회귀계수: {model.params[model.pvalues[1:].idxmin()]}')
```
