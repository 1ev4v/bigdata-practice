```python
# 2
import numpy as np
import pandas as pd
train_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/9_2_train.csv')
test_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/9_2_test.csv')

# print(train_df.info())
# print(train_df.head())
# print(test_df.info())

train_y = train_df['라벨']
train_X = train_df.drop(columns=['ID', '라벨'])
test_y = test_df['라벨']
test_X = test_df.drop(columns=['ID', '라벨'])

from sklearn.model_selection import train_test_split
train_X, valid_X, train_y, valid_y = train_test_split(train_X, train_y, test_size=0.3)
# print(train_X.shape, train_y.shape, valid_X.shape, valid_y.shape)

train_X_num = train_X.select_dtypes('number').columns.tolist()
valid_X_num = valid_X.select_dtypes('number').columns.tolist()
test_X_num = test_X.select_dtypes('number').columns.tolist()
train_X_cat = train_X.select_dtypes('object').columns.tolist()
valid_X_cat = valid_X.select_dtypes('object').columns.tolist()
test_X_cat = test_X.select_dtypes('object').columns.tolist()

from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
train_X_encoder = encoder.fit_transform(train_X[train_X_cat])
valid_X_encoder = encoder.transform(valid_X[valid_X_cat])
test_X_encoder = encoder.transform(test_X[test_X_cat])

train_X = np.concatenate([train_X[train_X_num], train_X_encoder], axis=1)
valid_X = np.concatenate([valid_X[valid_X_num], valid_X_encoder], axis=1)
test_X = np.concatenate([test_X[test_X_num], test_X_encoder], axis=1)

from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier()

rf.fit(train_X, train_y)
pred = rf.predict(valid_X)

from sklearn.metrics import f1_score
f1_score(valid_y, pred, average='macro')

pred2 = rf.predict(test_X)
pd.DataFrame(pred2, columns=['pred']).to_csv('result.csv', index=False)
```

```python
# 3-1
import numpy as np
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/9_3_1.csv')
# print(df.info())
# print(df.head())

train_df = df.iloc[:140, :]
test_df = df.iloc[140:, :]

train_y = train_df['design']
train_X = train_df.drop(columns=['design', 'id'])
train_df = train_df.drop(columns=['id'])
test_y = test_df['design']
test_X = test_df.drop(columns=['design', 'id'])

import statsmodels.api as sm
import statsmodels.formula.api as smf

model = smf.ols("design ~ tenure + f2 + f3 + f4 + f5", data=train_df).fit()
# print(model.summary())
print((model.params < 0.05).sum())

pred = model.predict(train_X)
# print(pred)

from scipy.stats import pearsonr
statistic, p_value = pearsonr(pred, train_y)
print(np.round(statistic, 2))

pred2 = model.predict(test_X)
from sklearn.metrics import root_mean_squared_error
print(np.round(root_mean_squared_error(test_y, pred2), 0))
```
