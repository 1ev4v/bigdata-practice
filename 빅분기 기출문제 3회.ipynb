```python
# 2
import numpy as np
import pandas as pd
train_X = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/3_2_trainX.csv')
train_y = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/3_2_trainy.csv')
test_X = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/3_2_testX.csv')

# print(train_X.head())
# print(train_X.info())
# print(train_y.head())
# print(train_y.info())
# print(test_X.head())
# print(test_X.info())

from sklearn.model_selection import train_test_split
train_X, valid_X, train_y, valid_y = train_test_split(train_X, train_y, test_size=0.3)

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

scale_train_X = scaler.fit_transform(train_X)
scale_valid_X = scaler.transform(valid_X)
scale_test_X = scaler.transform(test_X)

from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier()
rf.fit(scale_train_X, train_y)

pred1 = rf.predict(scale_valid_X)
from sklearn.metrics import f1_score
print(f1_score(valid_y, pred1))

pred = rf.predict(scale_test_X)
pred = pd.DataFrame(pred, columns=['pred']).to_csv('result.csv', index=False)
```

```python
# 3-2-1
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/3_3_2.csv')
# print(df.head())
# print(df.info())

df1 = pd.crosstab(df['gender'], df['club_membership'])

from scipy.stats import chi2_contingency
statistic, p_value, dof, expected = chi2_contingency(df1, correction=False)
print(round(statistic, 2))


# ------------------------------------------------------
# chi2은 crosstab을 통해 교차표가 있어야지 사용할 수 있다.
# 또한 chi2은 모델을 돌리고 나서 4개의 값을 받는다.

# correction에 대해서 알아보기!
# correction을 사용하라 말라라는 조건이 아마 문제에 나옴? 그걸보고 false, true를 결정하면 됨. 기본값은 true
```
