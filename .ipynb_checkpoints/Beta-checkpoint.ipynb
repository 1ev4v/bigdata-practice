```python
import pandas as pd
import numpy as np

np.random.seed(42)
n_samples = 100
X = np.random.randn(n_samples, 5)
y = 3 * X[:, 0] + 2 * X[:, 1] + X[:, 2] + np.random.randn(n_samples)
df = pd.DataFrame(X, columns=['var1', 'var2', 'var3', 'var4', 'var5'])
df['target'] = y

# print(df.head())

# 1번
# print(df.corr())
# print(df.corr()['target'].drop('target').idxmax())

# 2번
import statsmodels.formula.api as smf
model = smf.ols("target ~ var1 + var2 + var3 + var4 + var5", data=df).fit()
# dir(model)
# print(f'결정계수: {model.rsquared}')

# 3번
# print(model.summary())
print(f'변수이름: {model.pvalues.idxmax()}, p-value값: {model.pvalues[model.pvalues.idxmax()]}')
```

```python
import pandas as pd
import numpy as np
from sklearn.datasets import make_regression
import statsmodels.api as sm

X, y = make_regression(n_samples = 100, n_features = 3, noise = 0.1, random_state = 42)
df = pd.DataFrame(X, columns = [f'var{i}' for i in range(3)])
df['target'] = y

# print(df.head())

import statsmodels.formula.api as smf

model = smf.ols("target ~ var0 + var1 + var2", data=df).fit()
dir(model)
# print(model.conf_int())
# print(model.summary())
# print(f'유의확률이 가장 작은 변수의 회귀계수: {model.params[model.pvalues.idxmin()]}')

# print(f'해당 모델의 결정계수: {model.rsquared}')

test = pd.DataFrame({'const': [1.0],
                     'var0': [0.5],
                     'var1': [1.2],
                     'var2': [0.3]})
# test
rs = model.predict(test)
print(f'예측한 값: {rs[0]}')
```

```python
import pandas as pd
import numpy as np

np.random.seed(42)
n_samples = 210
X = np.random.randn(n_samples, 4)
y = (X[:, 0] + X[:, 1] * 0.5 + np.random.randn(n_samples) * 0.5 > 0).astype(int)
df = pd.DataFrame(X, columns=['weight', 'height', 'age', 'income'])
df['gender'] = y

# print(df.head())

import statsmodels.formula.api as smf

# dir(smf)
# help(smf.logit)
# model = smf.logit("gender ~ weight", data=df).fit()
# print(model.summary())
# print(f"몸무게에 대한 오즈비값: {np.exp(model.params['weight'])}")

import statsmodels.api as sm
# dir(sm.families)
# model = smf.glm("gender ~ weight + height + age + income", data=df, family=sm.families.Binomial()).fit()
# dir(model)
# print(f'residual deviance: {model.deviance.round(3)}')


import sklearn
from sklearn.model_selection import train_test_split
# help(sklearn.model_selection.train_test_split)
# dir(sklearn.model_selection)

train_df, test_df = train_test_split(df, test_size=90, train_size=10)
# print(train_df.shape)
# print(test_df.shape)
# train_y = train_df.pop('gender')
# print(train_y)
# print(train_df)
test_y = test_df.pop('gender')
test_df = test_df[['weight']]

# print(test_y)
# print(test_df)

from sklearn.metrics import accuracy_score
model = smf.glm("gender ~ weight", data=train_df, family=sm.families.Binomial())
y_pred = model.predict(test_df) > 0.5

error_rate = 1 - accuracy_score(test_y, y_pred)
print(error_rate)
```

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes

diabetes = load_diabetes(as_frame=True)
df = diabetes.frame
# print(df.head())

for i in range(len(df.target)):
    if df.target[i] > df.target.median():
        df.target[i] = 1
    else:
        df.target[i] = 0

import statsmodels.formula.api as smf
model = smf.logit("target ~ age + sex + bmi + bp", data=df).fit()
print(model.summary())
```
