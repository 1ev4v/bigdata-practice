```python
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/7_1_1.csv')
# print(df.head())
# print(df.info())
df['처리시간'] = df['처리시각'] - df['신고시각']
print(df.groupby('공장명')['처리시간'].mean().sort_values(ascending=True))
# print(help(pd.Series([1,2,3]).sort_values))
```

```python
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/7_1_2.csv')
# print(df.head())
# print(df.info())

# print(dir(df['STATION_ADDR1'].str.split))
df['구'] = df['STATION_ADDR1'].str.split(' ').str[1]
np.round(df.groupby('구')['dist'].mean()[['마포구','성동구']],3)
```
