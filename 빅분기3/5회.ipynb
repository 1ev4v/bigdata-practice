```python
# 2
import numpy as np
import pandas as pd
train_df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/5_2_train.csv')
test_df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/5_2_test.csv')
# print(train_df.info())
# print(train_df.head())
# print(test_df.info())

train_y=train_df['target']
train_X=train_df.drop(columns=['level_0', 'ID', 'target'])
# print(train_y.info())
# print(train_X.info())
test_y=test_df['target']
test_X=test_df.drop(columns=['level_0', 'ID', 'target'])

from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.metrics import roc_auc_score, get_scorer_names
from sklearn.ensemble import RandomForestClassifier
# print(dir(sklearn.ensemble))
# roc_auc

train_X, valid_X, train_y, valid_y=train_test_split(train_X, train_y, test_size=0.3)
# print(train_X.shape, train_y.shape, valid_X.shape, valid_y.shape)

rf=RandomForestClassifier()
rf.fit(train_X, train_y)
pred1=rf.predict_proba(valid_X)
print(roc_auc_score(valid_y, pred1[:,1]))

params={
    'max_depth': [10,20,30],
    'min_samples_split': [2,5,10]
}
cv=GridSearchCV(estimator=rf, param_grid=params, scoring='roc_auc', cv=3)
cv.fit(train_X, train_y)
pred2=cv.predict_proba(valid_X)
print(roc_auc_score(valid_y, pred2[:,1]))

pred3=rf.predict_proba(test_X)
print(roc_auc_score(test_y, pred3[:,1]))

pred4=cv.predict_proba(test_X)
print(roc_auc_score(test_y, pred4[:,1]))

pd.DataFrame(pred3[:,1], columns=['pred']).to_csv('result.csv', index=False)
```

```python
# 3-1
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/5_3_1.csv')
# print(df.info())
# print(df.head())

print(np.round(df['study_hours'].mean(),2))
print(np.round(df['study_hours'].std(),2))
print(np.round(df['exam_scores'].mean(),2))
print(np.round(df['exam_scores'].std(),2))

from scipy.stats import pearsonr
print(np.round(pearsonr(df['study_hours'], df['exam_scores']).statistic,3))
print(pearsonr(df['study_hours'], df['exam_scores']).pvalue < 0.05)
```

```python
# 3-2
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/5_3_2.csv')
# print(df.info())
# print(df.head())

print(np.round(df.groupby('campaign')['satisfaction_score'].mean().sum(), 2))
print(np.round(df.groupby('campaign')['satisfaction_score'].std().sum(), 2))

from scipy.stats import f_oneway
a=df[df['campaign']=='A']['satisfaction_score']
b=df[df['campaign']=='B']['satisfaction_score']
c=df[df['campaign']=='C']['satisfaction_score']
print(np.round(f_oneway(a,b,c).statistic, 3))
print(f_oneway(a,b,c).pvalue < 0.05)
```
