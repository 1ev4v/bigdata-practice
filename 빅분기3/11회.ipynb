```python
# 3-2
import numpy as np
import pandas as pd
df=pd.read_csv('C:/Users/user/OneDrive/바탕 화면/빅분기실기 기출 문제 및 데이터/데이터셋/11회_데이터셋/11.03.02.rental_property.csv')
# print(df.info())
# print(df.head())
import statsmodels.api as sm
import statsmodels.formula.api as smf
model=smf.ols('monthly_rent ~ area + floor + year_built + rooms + parking', data=df).fit()
# print(model.summary())
print(np.round(model.params['area'],3))
print(np.round(model.pvalues['parking'],3))
test_X=pd.DataFrame({'area': [75], 'floor': [12], 'year_built': [2015], 'rooms': [3], 'parking': [1]})
pred1=model.predict(test_X)
print(np.round(pred1[0],3))
```
