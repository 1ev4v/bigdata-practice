```python
# 1-3
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/3_1.csv')
# print(df.info())
# print(df.head())

up=df['MedianStudentsPerTeacher'].mean() + (1.5 * df['MedianStudentsPerTeacher'].std())
down=df['MedianStudentsPerTeacher'].mean() - (1.5 * df['MedianStudentsPerTeacher'].std())

df[(df['MedianStudentsPerTeacher'] >= up) | (df['MedianStudentsPerTeacher'] <= down)]['MedianStudentsPerTeacher'].count()
```

```python
# 2
import numpy as np
import pandas as pd
train_X=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/3_2_trainX.csv')
train_y=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/3_2_trainy.csv')
test_X=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/3_2_testX.csv')
# print(train_X.info())
# print(train_X.head())
# print(train_y.info())
# print(test_X.info())

from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.metrics import f1_score, get_scorer_names
from sklearn.ensemble import RandomForestClassifier
# print(get_scorer_names())
# f1

train_X, valid_X, train_y, valid_y=train_test_split(train_X, train_y, test_size=0.3)
# print(train_X.shape, train_y.shape, valid_X.shape, valid_y.shape)

rf=RandomForestClassifier()
rf.fit(train_X, train_y)
pred1=rf.predict(valid_X)
print(f1_score(valid_y, pred1))

params={
    'max_depth': [10,20,30],
    'min_samples_split': [2,5,10]
}
cv=GridSearchCV(estimator=rf, param_grid=params, scoring='f1', cv=3)
cv.fit(train_X, train_y)
pred2=cv.predict(valid_X)
print(f1_score(valid_y, pred2))

pred3=cv.predict(test_X)
pd.DataFrame(pred3,columns=['pred']).to_csv('result.csv', index=False)
```

```python
# 3-1
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/3_3_1.csv')
# print(df.info())
# print(df.head())

print(np.round(df['score'].mean(),2))
print(np.round(df['score'].std(),2))

from scipy.stats import ttest_1samp
print(np.round(ttest_1samp(df['score'], popmean=75).statistic, 2))
print(ttest_1samp(df['score'], popmean=75).pvalue < 0.05)
```

```python
# 3-2
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/3_3_2.csv')
# print(df.info())
# print(df.head())

df1=pd.crosstab(df['gender'], df['club_membership'])

from scipy.stats import chi2_contingency
print(np.round(chi2_contingency(df1, correction=False).statistic, 2))
print(chi2_contingency(df1, correction=False).pvalue < 0.05)
```
