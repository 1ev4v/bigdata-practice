```python
# 2
import numpy as np
import pandas as pd
train_df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/4_2_train.csv')
test_df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/4_2_test.csv')
# print(train_df.info())
# print(train_df.head())
# print(test_df.info())
# train_df['Attrition_Flag'].value_counts()
train_df['Attrition_Flag']=train_df['Attrition_Flag'].map({'Existing Customer': 0, 'Attrited Customer': 1})
test_df['Attrition_Flag']=test_df['Attrition_Flag'].map({'Existing Customer': 0, 'Attrited Customer': 1})

train_y=train_df['Attrition_Flag']
train_X=train_df.drop(columns=['ID', 'Attrition_Flag'])
test_y=test_df['Attrition_Flag']
test_X=test_df.drop(columns=['ID', 'Attrition_Flag'])
# print(train_X.info())
# print(train_y.info())
# print(test_X.info())
# print(test_y.info())

from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.metrics import roc_auc_score, get_scorer_names
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import OneHotEncoder
# print(dir(sklearn.preprocessing))
# roc_auc

train_X, valid_X, train_y, valid_y=train_test_split(train_X, train_y, test_size=0.3)
# print(train_X.shape, train_y.shape, valid_X.shape, valid_y.shape)

obj=train_X.select_dtypes('object').columns.tolist()
num=train_X.select_dtypes('number').columns.tolist()
encoder=OneHotEncoder(handle_unknown='ignore', sparse_output=False)
train_X_encoder=encoder.fit_transform(train_X[obj])
valid_X_encoder=encoder.transform(valid_X[obj])
test_X_encoder=encoder.transform(test_X[obj])

train_X=np.concatenate([train_X_encoder, train_X[num]], axis=1)
valid_X=np.concatenate([valid_X_encoder, valid_X[num]], axis=1)
test_X=np.concatenate([test_X_encoder, test_X[num]], axis=1)

rf=RandomForestClassifier()
rf.fit(train_X, train_y)
pred1=rf.predict_proba(valid_X)
print(roc_auc_score(valid_y, pred1[:,1]))

params={
    'max_depth': [10,20,30],
    'min_samples_split': [2,5,10]
}
cv=GridSearchCV(estimator=rf, param_grid=params, scoring='roc_auc', cv=3)
cv.fit(train_X, train_y)
pred2=cv.predict_proba(valid_X)
print(roc_auc_score(valid_y, pred2[:,1]))

pred3=cv.predict_proba(test_X)
print(roc_auc_score(test_y, pred3[:,1]))

pd.DataFrame(pred3[:,1], columns=['pred']).to_csv('result.csv', index=False)
```

```python
# 3-1
import numpy as np
import pandas as pd
df=pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/4_3_1.csv')
# print(df.info())
# print(df.head())

print(np.round(df[df['department']=='A']['hours_worked'].mean() + df[df['department']=='B']['hours_worked'].mean(), 2))
print(np.round(df[df['department']=='A']['hours_worked'].std() + df[df['department']=='B']['hours_worked'].std(), 2))

from scipy.stats import ttest_ind
print(np.round(ttest_ind(df[df['department']=='A']['hours_worked'], df[df['department']=='B']['hours_worked']).statistic, 2))
print(ttest_ind(df[df['department']=='A']['hours_worked'], df[df['department']=='B']['hours_worked']).pvalue.round(2) < 0.05)
```
