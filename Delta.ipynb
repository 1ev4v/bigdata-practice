```python
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_1_1.csv')

df.info()
print(df.groupby('대륙')['맥주소비량'].mean().idxmax())
print(df[df['대륙']==df.groupby('대륙')['맥주소비량'].mean().idxmax()].groupby('국가')['맥주소비량'].sum().sort_values(ascending=False).index[4])
print(df[df['국가']==df[df['대륙']==df.groupby('대륙')['맥주소비량'].mean().idxmax()].groupby('국가')['맥주소비량'].sum().sort_values(ascending=False).index[4]]['맥주소비량'].mean())
```

```python
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_1_2.csv')

print(df.info())
print(df.head())

df['관광객비율'] = df['관광'] / df[['관광', '사무', '공무', '유학', '기타']].sum(axis=1)
df['관광객비율'].sort_values(ascending=False).iloc[1]
```
