```python
import numpy as np
import pandas as pd

train_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_2_train.csv')
test_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_2_test.csv')

train_X = train_df.drop('count', axis=1)
train_y = train_df['count']
test_X = test_df.drop('count', axis=1)
test_y = test_df['count']

test_id = test_X['ID']

from sklearn.model_selection import train_test_split
train_X, valid_X, train_y, valid_y = train_test_split(train_X, train_y, test_size=0.3, random_state=1)
# print(train_X.shape, valid_X.shape, train_y.shape, valid_y.shape)

cat_X = train_X.select_dtypes('object').columns.to_list()
num_X = train_X.select_dtypes('number').columns.to_list()

from sklearn.preprocessing import OneHotEncoder
onehotencoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
one_train_X = onehotencoder.fit_transform(train_X[cat_X])
one_valid_X = onehotencoder.transform(valid_X[cat_X])
one_test_X = onehotencoder.transform(test_X[cat_X])

train_X1 = np.concatenate([one_train_X, train_X[num_X]], axis=1)
valid_X1 = np.concatenate([one_valid_X, valid_X[num_X]], axis=1)
test_X1 = np.concatenate([one_test_X, test_X[num_X]], axis=1)

from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(random_state=1)
rf.fit(train_X1, train_y)

pred = rf.predict(valid_X1)
pred1 = rf.predict(test_X1)

from sklearn.metrics import median_absolute_error
print(median_absolute_error(valid_y, pred))


rs = pd.DataFrame({'pred': pred1})
rs1 = pd.concat([test_id, rs], axis=1)
rs1.to_csv('result1.csv', index=False)

# test_df.info()
# data = pd.read_csv('result1.csv')
# data
```

```python
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_3_1.csv')

# print(df.info())
# print(df.head())

print(np.round(df['before'].mean(),2))
print(np.round(df['before'].std(),2))
print(np.round(df['after'].mean(),2))
print(np.round(df['after'].std(),2))
print('-----------') 

from scipy.stats import wilcoxon
statistic, pvalue = wilcoxon(df['before'], df['after'])
print(np.round(statistic,2))
if pvalue < 0.05:
    print('기각')
else:
    print('채택')
```

```python
import numpy as np
import pandas as pd

train_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_3_2_train.csv')
test_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/8_3_2_test.csv')

# print(train_df.info())
# print(train_df.head())

import statsmodels.formula.api as smf
model = smf.ols("productivity ~ hours + age + experience", data=train_df).fit()
# print(model.summary())
print(model.params[1:].idxmax())
print((np.round(model.pvalues[1:],4) < 0.05).count())

test_X = test_df.drop('productivity', axis=1)
test_y = test_df['productivity']
pred = model.predict(test_X)

from sklearn.metrics import r2_score
print(r2_score(test_y,pred))
```
