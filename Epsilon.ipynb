```python
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/9_1_3.csv')

# print(df.info())
# print(df.head())

df['평균만족도'] = df['평균만족도'].fillna(np.round(df['평균만족도'].mean(),6))
df['근속연수'] = df['근속연수'].fillna(np.floor(df.groupby(['부서', '등급'])['근속연수'].transform('mean')))

A = df[(df['부서'] == 'HR') & (df['등급'] == 'A')]['근속연수'].mean()
print(A)
B = df[(df['부서'] == 'Sales') & (df['등급']=='B')]['교육참가횟수'].mean()
print(B)
print(np.round((A+B),4))
```

```python
import numpy as np
import pandas as pd

train_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/9_2_train.csv')
test_df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/9_2_test.csv')

print(train_df.info())
print(train_df.head())

train_X = train_df.drop(['라벨', 'ID'], axis=1)
train_y = train_df['라벨']
test_X = test_df.drop(['라벨', 'ID'], axis=1)
test_y = test_df['라벨']

from sklearn.model_selection import train_test_split
train_X, valid_X, train_y, valid_y = train_test_split(train_X, train_y, test_size=0.3)

cat_train_X = train_X.select_dtypes('object').columns.to_list()
num_train_X = train_X.select_dtypes('number').columns.to_list()

from sklearn.preprocessing import OneHotEncoder
onehotencoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
one_cat_train_X = onehotencoder.fit_transform(train_X[cat_train_X])
one_cat_valid_X = onehotencoder.transform(valid_X[cat_train_X])
one_cat_test_X = onehotencoder.transform(test_X[cat_train_X])

train_X1 = np.concatenate([one_cat_train_X, train_X[num_train_X]], axis=1)
valid_X1 = np.concatenate([one_cat_valid_X, valid_X[num_train_X]], axis=1)
test_X1 = np.concatenate([one_cat_test_X, test_X[num_train_X]], axis=1)

from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier()
rf.fit(train_X1, train_y)

pred1 = rf.predict(valid_X1)

from sklearn.metrics import f1_score
rs = f1_score(valid_y, pred1, average='macro')
print(rs)

pred2 = rf.predict(test_X1)
pred2 = pd.DataFrame({'pred': pred2})
pred2.to_csv('result.csv', index=False)
```
