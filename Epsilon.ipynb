```python
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/9_1_1.csv')

print(df.info())
print(df.head())

df['총액'] = df['금액1'] + df['금액2']
df
df1 = df.groupby(['year', '지역코드', 'gender'])['총액'].sum().reset_index()
df2 = df1.pivot_table(index=['year', '지역코드'], columns='gender', values='총액', fill_value=0)
df2['sum'] = abs(df2.iloc[:,0] - df2.iloc[:,1])
print(df2[df2['sum']==df2['sum'].sort_values(ascending=False).iloc[0]].index.values[0][1])
```

```python
import numpy as np
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/9_1_2.csv')

# print(df.info())
# print(df.head())

df['구분1'] = df['구분'].str.split('건수').str[0]
df['범죄합'] = df[['범죄유형1', '범죄유형2', '범죄유형3', '범죄유형4', '범죄유형5', '범죄유형6', '범죄유형7', '범죄유형8', '범죄유형9', '범죄유형10']].sum(axis=1)
df
# df[df['구분1']=='검거']
# df[df['구분1']=='발생']
rs = df.groupby(['연도', '구분1'])['범죄합'].sum().reset_index()
pivot = rs.pivot(index='연도', columns='구분1', values='범죄합')
pivot['검거율'] = pivot['검거'] / pivot['발생']
print(pivot[pivot['검거율']==pivot['검거율'].sort_values(ascending=False).iloc[0]]['검거'].values[0])
# print(pivot[pivot['검거율']==pivot['검거율'].sort_values(ascending=False).iloc[0]]['검거'])
# (df.groupby('연도')['범죄합'].sum()).sum()
```
