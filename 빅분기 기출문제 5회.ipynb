```python
# 1-1   x
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/5_1_1.csv')
# print(df.head())
# print(df.info())

df['conventional'] = df['conventional'].str.replace(r'[^0-9]', '', regex=True)
df['conventional'] = pd.to_numeric(df['conventional'], errors='coerce')
# df['conventional']
df['conventional'] = df['conventional'].fillna(0).astype(int)
df['conventional'] = df['conventional'].astype(int)

rs = round(df['conventional'].quantile(0.75) - df['conventional'].quantile(0.25), 1)
print(rs)
```

```python
# 1-1   o
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/YoungjinBD/data/main/exam/5_1_1.csv')
# print(df.head())
# print(df.info())

cond = df['conventional'].str.contains(r'[^0-9]', regex=True)
df = df[~cond]
df['conventional'] = df['conventional'].astype(int)
# print(df.info())

rs = round(df['conventional'].quantile(0.75) - df['conventional'].quantile(0.25), 0)
print(rs)
```
